```julia
using IntervalArithmetic
```

```julia
par_noise = 0.2971290840221:0.00000000000001:0.2971290840222
lambdas = MultipleExperiments(interval(3.5), interval(1), 128, par_noise)
```

```julia
function T(x; α, β)
    return β-(1+β)*abs(x)^(α)
end
```

```julia
function bc(x)
    return x - 2*floor((x+1)/2)
end
```

```julia
noise_vector = randn(BigFloat, 1024)*0.2971290840221
```

```julia
function orbit_given_noise(T, x, noise_vector)
    orbit = zeros(BigFloat, length(noise_vector))
    for i in 1:length(orbit)
        x = bc(T(x)+noise_vector[i])
        orbit[i] = x
    end
    return orbit
end
```

```julia
x0 = BigFloat(-0.1)
y0 = BigFloat(0.9)

T(x) = T(x; α = 3.5, β = 1.0)
```

```julia
orb_x = orbit_given_noise(T, x0, noise_vector)
orb_y = orbit_given_noise(T, y0, noise_vector)
```

```julia
dist = abs.(orb_x-orb_y)
```

```julia
using Plots, LaTeXStrings
```

```julia
plot(dist, label= L"\alpha= 3.5, \beta = 1.0, \sigma = 0.2971290840221")
```

```julia
savefig("intermittent.png")
```
